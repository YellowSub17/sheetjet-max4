```python
import matplotlib.pyplot as plt
import numpy as np
import h5py
import hdf5plugin
```

```python
PROC_DIR = '/home/pat/max4/sheetjet-max4/data/proc'
```

```python
def check_run(group, subgroup, n):
    with h5py.File(f'{PROC_DIR}/{group}/{group}-{subgroup}_{n}_proc.h5') as f:
        
        mask = f['mask'][...]
        I = f['powder/I'][...]
        q = f['powder/q'][...]
        counts = f['powder/counts'][...]
        std = f['powder/std'][...]
        mean = f['mean'][...]

    fig, axes, = plt.subplots(1,1)
    axes.set_title(f'{group}-{subgroup}_{n}')
    rows = I[::10, :]                              # one line per row
    cmap = plt.cm.viridis                          # any colormap works
    colors = cmap(np.linspace(0, 1, rows.shape[0]))
    for color, row in zip(colors, rows):
        axes.plot(q, row, color=color)
    fig, axes = plt.subplots(1,2)
    
    axes[0].imshow(mean, vmin=0, vmax=np.median(mean)*4)
    axes[0].set_title(f'mean')
    axes[1].imshow(mask)
    axes[1].set_title(f'mask')


        
```

```python
check_run('gas_background', 'gas_background', 1)
check_run('celluloseNC_100x', 'celluloseNC_100x', 6)
check_run('celluloseNC_100x', 'celluloseNC_100x', 9)
```

```python
check_run('Liposome_POPC_1000x', 'longaxis', 4)
check_run('Liposome_POPC_1000x', 'longaxis', 2)
check_run('Liposome_POPC_20x_water', 'longaxis', 2)
```
